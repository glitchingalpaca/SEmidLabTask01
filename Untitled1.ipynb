```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# ==========================================
# S301 RELATIVISTIC OBSERVER - VERSION 1.0
# Stage 1: Newtonian Orbital Simulation
# ==========================================

# 1. Physical constants
G = 6.67430e-11
C = 299792458
M_SUN = 1.98847e30
AU = 1.495978707e11

# 2. Sagittarius A* and S301 parameters
mass = 4.297e6 * M_SUN

a = 83 * 8.277       # Semi-major axis in AU
e = 0.9832            # Orbital eccentricity
period = 8.7          # Orbital period in years

# 3. Calculate orbital characteristics
rp = a * (1 - e)
ra = a * (1 + e)

mu = G * mass

# Vis-viva equation: orbital velocity
def orbital_speed(r):
    r_m = r * AU
    return np.sqrt(mu * (2 / r_m - 1 / (a * AU)))

v_peri = orbital_speed(rp)
v_apo = orbital_speed(ra)

# 4. Solve Kepler's equation
def solve_kepler(M, eccentricity):
    E = M.copy()

    for i in range(20):
        E -= (
            E - eccentricity * np.sin(E) - M
        ) / (
            1 - eccentricity * np.cos(E)
        )

    return E

# 5. Generate orbital positions
frames = 160

mean_anomaly = np.linspace(
    0, 2 * np.pi, frames
)

E = solve_kepler(mean_anomaly, e)

# Position in orbital plane
x = a * (np.cos(E) - e)
y = a * np.sqrt(1 - e**2) * np.sin(E)

# Distance from black hole
r = a * (1 - e * np.cos(E))

# Calculate velocity at every position
velocity = np.array([
    orbital_speed(distance) for distance in r
])

# 6. Create the visualization
fig = plt.figure(figsize=(12, 7))
fig.patch.set_facecolor("#080b12")

ax = fig.add_subplot(121)
info = fig.add_subplot(122)

ax.set_facecolor("#080b12")
info.set_facecolor("#080b12")

# Draw orbital path
ax.plot(
    x, y,
    color="cyan",
    alpha=0.45,
    linewidth=1.5,
    linestyle="--"
)

# Draw Sagittarius A*
black_hole = plt.Circle(
    (0, 0), 8,
    color="black",
    ec="orange",
    linewidth=2,
    zorder=5
)

ax.add_patch(black_hole)

ax.text(
    0, -35, "Sgr A*",
    color="white",
    ha="center"
)

# Draw S301
star, = ax.plot(
    [], [],
    marker="*",
    color="yellow",
    markersize=14,
    linestyle="None",
    zorder=10
)

# Draw trail
trail, = ax.plot(
    [], [],
    color="orange",
    linewidth=1.5,
    alpha=0.8
)

ax.set_xlim(-800, 800)
ax.set_ylim(-800, 800)
ax.set_aspect("equal")
ax.set_xlabel("Distance (AU)", color="white")
ax.set_ylabel("Distance (AU)", color="white")
ax.tick_params(colors="white")
ax.set_title(
    "S301 ORBITAL VIEW",
    color="white",
    fontsize=14
)

# 7. Information panel
info.axis("off")

info.set_title(
    "MISSION TELEMETRY",
    color="cyan",
    fontsize=16,
    pad=20
)

telemetry = info.text(
    0.05, 0.85, "",
    transform=info.transAxes,
    color="white",
    fontsize=12,
    linespacing=2.0,
    verticalalignment="top"
)

# 8. Animation function
def animate(i):

    star.set_data([x[i]], [y[i]])

    trail.set_data(
        x[:i+1],
        y[:i+1]
    )

    telemetry.set_text(
        f"""
        STAR: S301

        Orbital period: {period:.2f} years

        Eccentricity: {e:.4f}

        Distance from Sgr A*:
        {r[i]:,.2f} AU

        Orbital velocity:
        {velocity[i]/1000:,.0f} km/s

        Velocity as fraction of light:
        {velocity[i]/C:.4f} c

        Closest approach:
        {rp:.2f} AU

        Farthest distance:
        {ra:.2f} AU

        Simulation time:
        {mean_anomaly[i]/(2*np.pi)*period:.2f} years
        """
    )

    return star, trail, telemetry

# 9. Run animation
animation = FuncAnimation(
    fig,
    animate,
    frames=frames,
    interval=60,
    blit=False
)

plt.close(fig)

display(HTML(animation.to_jshtml()))


# STAGE 2: SPECIAL RELATIVITY
# Experiment 1: Time dilation

C = 299792458

def lorentz_factor(velocity):
    beta = velocity / C
    return 1 / np.sqrt(1 - beta**2)

# S301's orbital speeds
gamma_peri = lorentz_factor(v_peri)
gamma_apo = lorentz_factor(v_apo)

# Imagine 1 year passes in our reference frame
earth_time = 1.0

s301_time_peri = earth_time / gamma_peri
s301_time_apo = earth_time / gamma_apo

print("S301 RELATIVISTIC CLOCK")
print("-----------------------")

print(f"Periapsis speed: {v_peri/1000:.2f} km/s")
print(f"Periapsis Lorentz factor: {gamma_peri:.8f}")
print(f"Time experienced by S301: {s301_time_peri:.8f} years")

print()

print(f"Apoapsis speed: {v_apo/1000:.2f} km/s")
print(f"Apoapsis Lorentz factor: {gamma_apo:.8f}")
print(f"Time experienced by S301: {s301_time_apo:.8f} years")

print()

print("Difference after one reference-frame year:")
print(f"{(earth_time-s301_time_peri)*86400:.2f} seconds at periapsis")



# STAGE 2: RELATIVISTIC DOPPLER EFFECT
# Experiment 2: Visible light colour shift

import numpy as np
import matplotlib.pyplot as plt

C = 299792458

# Hypothetical green starlight
original_wavelength = 550  # nanometers

def doppler_factor(velocity, angle):
    beta = velocity / C
    gamma = 1 / np.sqrt(1 - beta**2)
    return 1 / (gamma * (1 - beta * np.cos(angle)))

# Convert wavelength into an approximate visible colour
def wavelength_to_rgb(wavelength):
    if wavelength < 380 or wavelength > 750:
        return (0.15, 0.15, 0.15)

    if wavelength < 440:
        return (0.5, 0, 1)
    elif wavelength < 490:
        return (0, 0, 1)
    elif wavelength < 510:
        return (0, 1, 1)
    elif wavelength < 580:
        return (0, 1, 0)
    elif wavelength < 645:
        return (1, 1, 0)
    else:
        return (1, 0, 0)

# S301's periapsis speed
velocity = v_peri

# Photons arriving from ahead and behind
angles = {
    "Ahead": 0,
    "Behind": np.pi
}

fig, ax = plt.subplots(figsize=(10, 4))
fig.patch.set_facecolor("#080b12")
ax.set_facecolor("#080b12")

for label, angle in angles.items():
    D = doppler_factor(velocity, angle)
    observed = original_wavelength / D

    color = wavelength_to_rgb(observed)

    ax.scatter(
        label,
        1,
        s=2500,
        color=color,
        edgecolors="white",
        linewidths=1.5
    )

    ax.text(
        label,
        0.65,
        f"{observed:.1f} nm\nD = {D:.4f}",
        color="white",
        ha="center",
        fontsize=11
    )

ax.set_ylim(0.4, 1.5)
ax.set_title(
    "S301 — RELATIVISTIC STARLIGHT",
    color="cyan",
    fontsize=14
)
ax.set_yticks([])
ax.tick_params(colors="white")
ax.spines[:].set_visible(False)

plt.show()

print(f"Original wavelength: {original_wavelength} nm")
print(f"S301 velocity: {velocity/1000:.0f} km/s")



# STAGE 2: RELATIVISTIC ABERRATION
# Experiment 3: Full sky map

import numpy as np
import matplotlib.pyplot as plt

C = 299792458

# Generate random stars across the celestial sphere
np.random.seed(42)

number_of_stars = 1200

z = np.random.uniform(-1, 1, number_of_stars)
phi = np.random.uniform(0, 2*np.pi, number_of_stars)

radius = np.sqrt(1-z**2)

# Original photon directions
nx = radius * np.cos(phi)
ny = radius * np.sin(phi)
nz = z

directions = np.column_stack((nx, ny, nz))

# S301 moves along the positive x direction
velocity = v_peri
beta = velocity / C
gamma = 1 / np.sqrt(1-beta**2)

beta_vector = np.array([beta, 0, 0])

# Lorentz transformation of photon directions
dot_product = directions @ beta_vector

coefficient = (
    (gamma - 1) * dot_product / beta**2 - gamma
)

new_directions = (
    directions + coefficient[:, None] * beta_vector
) / (
    gamma * (1 - dot_product)[:, None]
)

# Convert directions into sky-map coordinates
def sky_coordinates(d):
    longitude = np.arctan2(d[:, 1], d[:, 0])
    latitude = np.arcsin(np.clip(d[:, 2], -1, 1))

    return np.degrees(longitude), np.degrees(latitude)

old_lon, old_lat = sky_coordinates(directions)
new_lon, new_lat = sky_coordinates(new_directions)

# Plot both skies
fig, axes = plt.subplots(
    2, 1,
    figsize=(12, 9),
    subplot_kw={"projection": "mollweide"}
)

fig.patch.set_facecolor("#080b12")

for ax in axes:
    ax.set_facecolor("#080b12")
    ax.grid(color="gray", alpha=0.3)
    ax.tick_params(colors="white")

axes[0].scatter(
    np.radians(old_lon),
    np.radians(old_lat),
    s=3,
    color="white"
)

axes[0].set_title(
    "STATIONARY OBSERVER",
    color="cyan",
    pad=15
)

axes[1].scatter(
    np.radians(new_lon),
    np.radians(new_lat),
    s=3,
    color="orange"
)

axes[1].set_title(
    "OBSERVER MOVING WITH S301",
    color="cyan",
    pad=15
)

plt.tight_layout()
plt.show()

print(f"S301 speed: {velocity/1000:.0f} km/s")
print(f"Speed of light fraction: {beta:.4f}")
print(f"Lorentz factor: {gamma:.6f}")


```
